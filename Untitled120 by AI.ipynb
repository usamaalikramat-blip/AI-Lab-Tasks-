```python
import re
WORD = re.compile(r"\w+(?:['’]\w+)*")

texts = [
    "Hello, world! Wait... what?",
    "I don't think it's ready.",
    "Write to ali.khan@example.com today",
    "I enrolled in CS4944 this term",
    "",
    "میں اسکول جاتا ہوں۔",
]

for text in texts:
    print(repr(text))            
    print(text.split())          
    print(WORD.findall(text))
    print()

english = [texts[1], texts[2]]
try:
    from transformers import AutoTokenizer
    tok = AutoTokenizer.from_pretrained("gpt2")
    for s in english:
        print(s, "->", tok.tokenize(s))
except Exception:
    pass                         
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

faq = {
    "what are the library hours": ("Open 8am to 8pm.", "Library Handbook"),
    "how do i reset my password": ("Click 'Forgot password'.", "IT Help Desk"),
}
questions = list(faq)
tfidf = TfidfVectorizer().fit(questions)

while True:
    q = input("You: ").lower().strip()
    if q == "quit":
        break
    if q == "":
        print("Bot: Please type something.")
    elif q in faq:
        print("Bot: %s (Source: %s)" % faq[q])
    else:
        score = cosine_similarity(tfidf.transform([q]), tfidf.transform(questions))[0]
        best = questions[score.argmax()]
        if score.max() < 0.2:
            print("Bot: Sorry, I don't know that.")
        elif input(f"Bot: Did you mean '{best}'? (yes/no) ") == "yes":
            print("Bot: %s (Source: %s)" % faq[best])
```

```python
import json
import os
import re
import subprocess
import sys
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
BUDGET = 40
DOCS = {
    "refund": {
        "text": (
            "Northwind Software sells licences for its design tools through an online store, "
            "and every purchase is covered by the refund policy described here. "
            "Customers must request a refund within 30 days of the purchase date. "
            "The request is made through the account page, where the order number and a short reason are entered. "
            "Refunds are returned to the original payment method and usually appear within five to ten working days, "
            "although some banks take longer to post the credit. "
            "However, a licence that has already been activated cannot be refunded unless the product fails to run "
            "on supported hardware after the support team has tried to fix it. "
            "Team administrators who bought several seats may refund unused seats separately from activated ones. "
            "Gift purchases follow the same schedule, and the refund goes to the buyer rather than the recipient. "
            "Support staff review each request in the order received, and they may ask for screenshots or log files "
            "when a technical fault is the stated reason. "
            "Customers who are unhappy with a decision can ask for a second review by replying to the confirmation "
            "email, and a different team member will then read the case from the beginning."
        ),
        "rule": "Customers must request a refund within 30 days of the purchase date.",
        "exception": "However, a licence that has already been activated cannot be refunded unless the product "
                     "fails to run on supported hardware after the support team has tried to fix it.",
    },
    "lab": {
        "text": (
            "The chemistry teaching laboratory is shared by several classes, so a few simple safety habits apply to "
            "everybody who enters the building. "
            "Visitors must wear safety goggles at all times inside the laboratory. "
            "Goggles are kept in the blue cabinet next to the door and should be cleaned with the wipes provided "
            "before they are put back. "
            "Long hair must be tied back, open shoes are not allowed, and coats should be hung in the corridor "
            "instead of being left on the benches. "
            "The only exception is the reading room at the far end, where goggles may be removed because no "
            "experiments are permitted there. "
            "Food and drink stay outside the laboratory, and anyone who spills a chemical, however small the amount "
            "seems, must tell the supervisor immediately so that the area can be cleaned correctly and nobody else "
            "touches the surface by accident. "
            "Fire exits are marked in green and must never be blocked by boxes or chairs. "
            "The supervisor keeps a log of every incident, and the log is reviewed at the end of each month by the "
            "department safety committee."
        ),
        "rule": "Visitors must wear safety goggles at all times inside the laboratory.",
        "exception": "The only exception is the reading room at the far end, where goggles may be removed because "
                     "no experiments are permitted there.",
    },
    "travel": {
        "text": (
            "This note explains how employees of the regional office book and claim business travel during the year. "
            "Employees must book flights at least 14 days before departure. "
            "Fares are compared on the internal booking tool, and the cheapest suitable option should be chosen, "
            "e.g. an economy seat on a direct route when one is available. "
            "Dr. Rao from finance checks each claim against the booking record before payment is released. "
            "Unless a client meeting is scheduled with less than 14 days notice, in which case the manager may "
            "approve a late booking, the standard deadline applies without change. "
            "Hotel stays are capped at a nightly rate that depends on the city, and receipts must be uploaded "
            "within a week of returning to the office. "
            "When a trip involves several cities, several clients and a long chain of connecting flights that cannot "
            "be arranged through the normal tool, the employee should send the full plan to the travel desk, which "
            "will book everything on their behalf and report the total cost back to the manager. "
            "Meals are reimbursed up to a daily limit, and alcohol is never covered."
        ),
        "rule": "Employees must book flights at least 14 days before departure.",
        "exception": "Unless a client meeting is scheduled with less than 14 days notice, in which case the "
                     "manager may approve a late booking, the standard deadline applies without change.",
    },
}


def words_of(text):
    return [(m.start(), m.end()) for m in re.finditer(r"\S+", text)]
def make_chunk(text, doc_id, method, n, start, end):
    return {
        "id": f"{doc_id}:{method}:{n:02d}",
        "doc_id": doc_id,
        "method": method,
        "start": start,
        "end": end,
        "n_words": len(text[start:end].split()),
        "text": text[start:end],
    }
def fixed_windows(text, doc_id, size=BUDGET, overlap=0):
    if size <= 0:
        raise ValueError("size must be positive")
    if overlap < 0 or overlap >= size:
        raise ValueError("overlap must be >= 0 and smaller than size")
    spans = words_of(text)
    chunks = []
    step = size - overlap
    i = 0
    while i < len(spans):
        j = min(i + size, len(spans))
        chunks.append(make_chunk(text, doc_id, f"win{overlap}", len(chunks),
                                 spans[i][0], spans[j - 1][1]))
        if j == len(spans):
            break
        i += step
    return chunks
def split_sentences(text):
    """Naive: split after . ! ? followed by whitespace. Returns (start, end) spans."""
    spans = []
    start = 0
    for m in re.finditer(r"(?<=[.!?])\s+", text):
        spans.append((start, m.start()))
        start = m.end()
    if start < len(text.rstrip()):
        spans.append((start, len(text.rstrip())))
    return [(a, b) for a, b in spans if text[a:b].strip()]
def sentence_pack(text, doc_id, budget=BUDGET):
    if budget <= 0:
        raise ValueError("budget must be positive")
    chunks = []
    cur = None  
def flush():
        nonlocal cur
        if cur:
            chunks.append(make_chunk(text, doc_id, "sent", len(chunks), cur[0], cur[1]))
            cur = None

    for a, b in split_sentences(text):
        n = len(text[a:b].split())
        if n > budget:
            # sentence is bigger than the budget: hard split by words
            flush()
            sp = [(a + s, a + e) for s, e in words_of(text[a:b])]
            for k in range(0, len(sp), budget):
                part = sp[k:k + budget]
                chunks.append(make_chunk(text, doc_id, "sent", len(chunks),
                                         part[0][0], part[-1][1]))
            continue
        if cur and cur[2] + n > budget:
            flush()
        if cur:
            cur[1] = b
            cur[2] += n
        else:
            cur = [a, b, n]
    flush()
    return chunks
def survives(chunks, text, phrase):
    a = text.find(phrase)
    assert a >= 0, "labeled phrase not found in document"
    b = a + len(phrase)
    return any(c["start"] <= a and b <= c["end"] for c in chunks)
def stats(chunks, text):
    sizes = [c["n_words"] for c in chunks]
    if not sizes:
        return 0, 0, 0, 0.0, 0
    repeated = sum(sizes) - len(text.split())
    return len(sizes), min(sizes), max(sizes), sum(sizes) / len(sizes), repeated
def section(title):
    print("\n" + "=" * 70)
    print(title)
    print("=" * 70)
def main():
    section("1. Document sizes")
    for d, v in DOCS.items():
        n = len(v["text"].split())
        longest = max(len(v["text"][a:b].split()) for a, b in split_sentences(v["text"]))
        print(f"{d:8s} words={n}  longest sentence={longest} words")
        assert 150 <= n <= 250, f"{d} must be 150-250 words"

    section("2. Comparison of chunkers (40-word budget)")
    all_chunks = {"win0": [], "win10": [], "sent": []}
    print(f"{'doc':8s} {'method':7s} {'n':>3s} {'min':>4s} {'max':>4s} {'avg':>6s} {'repeat':>7s}  "
          f"rule  exception  both-together")
    for d, v in DOCS.items():
        t = v["text"]
        sets = {
            "win0": fixed_windows(t, d, BUDGET, 0),
            "win10": fixed_windows(t, d, BUDGET, 10),
            "sent": sentence_pack(t, d, BUDGET),
        }
        for name, ch in sets.items():
            all_chunks[name].extend(ch)
            n, lo, hi, avg, rep = stats(ch, t)
            r = survives(ch, t, v["rule"])
            e = survives(ch, t, v["exception"])
            a = t.find(v["rule"])
            b = t.find(v["exception"]) + len(v["exception"])
            lo_, hi_ = min(a, t.find(v["exception"])), max(b, a + len(v["rule"]))
            both = any(c["start"] <= lo_ and hi_ <= c["end"] for c in ch)
            print(f"{d:8s} {name:7s} {n:3d} {lo:4d} {hi:4d} {avg:6.1f} {rep:7d}  "
                  f"{str(r):5s} {str(e):10s} {both}")

    section("3. Offset check (text[start:end] == stored text) and IDs")
    for name, ch in all_chunks.items():
        ok = all(DOCS[c["doc_id"]]["text"][c["start"]:c["end"]] == c["text"] for c in ch)
        ids = [c["id"] for c in ch]
        print(f"{name:6s} chunks={len(ch):2d} offsets_ok={ok} unique_ids={len(set(ids)) == len(ids)} "
              f"first_id={ids[0]}")

    section("4. Edge case tests")
    print("empty text, windows      :", fixed_windows("", "e", BUDGET, 0))
    print("empty text, sentences    :", sentence_pack("", "e", BUDGET))
    short = "This is a short document. It is below the budget."
    s1 = fixed_windows(short, "short", BUDGET, 10)
    s2 = sentence_pack(short, "short", BUDGET)
    print("short doc windows        :", len(s1), "chunk(s),", s1[0]["n_words"], "words")
    print("short doc sentences      :", len(s2), "chunk(s),", s2[0]["n_words"], "words")
    for bad in (-1, 40, 55):
        try:
            fixed_windows("some text here", "bad", BUDGET, bad)
            print(f"overlap={bad}: NOT rejected (bug)")
        except ValueError as err:
            print(f"overlap={bad}: rejected -> {err}")
    t = DOCS["travel"]["text"]
    packed = sentence_pack(t, "travel", BUDGET)
    long_start = t.find("When a trip involves")
    pieces = [c for c in packed if c["start"] >= long_start and c["end"] <= t.find("Meals are")]
    print(f"sentence longer than budget: split into {len(pieces)} pieces of "
          f"{[c['n_words'] for c in pieces]} words, all <= {BUDGET}: "
          f"{all(c['n_words'] <= BUDGET for c in packed)}")

    section("5. Where sentence splitting fails")
    tricky = ("Dr. Rao met Mr. Khan at 3.30 p.m. in the U.S. office, e.g. to discuss v2.1. "
              "He said \"Stop!\" and left.")
    print("text:", tricky)
    for a, b in split_sentences(tricky):
        print("  piece ->", repr(tricky[a:b]))
    print("""
The splitter cuts after every . ! ? followed by a space, so it fails when:
 - abbreviations end in a full stop (Dr., Mr., e.g., p.m., U.S.) and the text breaks mid-sentence;
 - decimals or versions (3.30, v2.1) usually survive only because no space follows the dot;
 - a quote or bracket follows the punctuation, e.g. "Stop!" and left, is split from its sentence;
 - a sentence has no final punctuation (headings, bullet points), so two lines merge into one;
 - very long sentences stay as one piece and need a hard word split.
In the travel document, "Dr. Rao" is split in two, which is visible in the sentence chunks below.""")
    for c in sentence_pack(t, "travel", BUDGET):
        if "Dr." in c["text"] or c["text"].startswith("Rao"):
            print("  ", c["id"], "->", repr(c["text"][:70]))

    section("6. Guided scripts")
    records = [
        {"id": c["id"], "doc_id": c["doc_id"], "start": c["start"], "end": c["end"], "text": c["text"]}
        for c in all_chunks["sent"]
    ]
    with open("chunk_records.json", "w", encoding="utf-8") as f:
        json.dump(records, f, indent=2)
    print("saved chunk_records.json with", len(records), "records")
    for script in ("guided_embeddings.py", "guided_suggestions.py"):
        if not os.path.exists(script):
            print(f"{script}: file not found in {os.getcwd()}, not run")
            continue
        print(f"--- running {script}")
        res = subprocess.run([sys.executable, script], capture_output=True, text=True)
        print(res.stdout)
        if res.returncode != 0:
            print("exit code", res.returncode)
            print(res.stderr)

    section("7. TF-IDF retrieval over sentence-packed chunks")
    queries = [
        "How long do I have to ask for my money back after buying?",
        "Do I need to wear eye protection in the reading room?",
        "Can a late flight booking be allowed if the client call was arranged at short notice?",
    ]
    vec = TfidfVectorizer(stop_words="english")
    matrix = vec.fit_transform([r["text"] for r in records])
    for q in queries:
        sims = cosine_similarity(vec.transform([q]), matrix)[0]
        order = sims.argsort()[::-1][:2]
        print("\nQUERY:", q)
        for rank, i in enumerate(order, 1):
            r = records[i]
            original = DOCS[r["doc_id"]]["text"][r["start"]:r["end"]]
            print(f"  #{rank} {r['id']}  score={sims[i]:.3f}  chars {r['start']}-{r['end']}")
            print("     evidence:", original)


if __name__ == "__main__":
    main()
```
