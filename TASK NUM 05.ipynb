```python
def minimax(node, is_max, name="A", depth=0):
    indent = "    " * depth
    if isinstance(node, int):             
        print(f"{indent}{name} is a leaf -> {node}")
        return node

    player = "MAX" if is_max else "MIN"
    print(f"{indent}{name} ({player}) looks at its children...")

    values = []
    for i, child in enumerate(node):
        child_name = name + str(i + 1)     
        values.append(minimax(child, not is_max, child_name, depth + 1))

    result = max(values) if is_max else min(values)
    print(f"{indent}{name} ({player}) picks {result} from {values}")
    return result
tree = [[3, 5], [2, 9], [0, 7]]
best = minimax(tree, is_max=True)
print("\nFinal Minimax value at the root:", best)
```
